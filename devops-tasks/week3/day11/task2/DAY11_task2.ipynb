```python

import pandas as pd

df = pd.read_csv("sales_dirty.csv")

print("First 5 records:")
print(df.head(5))

print("\nMissing values using isnull():")
print(df.isnull())

print("\nMissing values using isna():")
print(df.isna())

print("\nMissing value count:")
print(df.isnull().sum())

df_clean = df.dropna()
print("\nDataset after dropna():")
print(df_clean)

print("\nUnitPrice median:", df["UnitPrice"].median())
print("UnitPrice mean:", df["UnitPrice"].mean())

df["UnitPrice"] = df["UnitPrice"].fillna(
    df["UnitPrice"].median()
)

print("\nUnitPrice after filling missing values:")
print(df[["Product", "UnitPrice"]])

df["Quantity"] = pd.to_numeric(
    df["Quantity"], errors="coerce"
)

df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)

print("\nFinal dataset:")
print(df)
```
