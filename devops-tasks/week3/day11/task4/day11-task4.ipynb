```python
import pandas as pd

df = pd.read_csv("sales_dirty.csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

display(df)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())

print("\nDuplicate rows:")
display(df[df.duplicated()])
```

```python
print(
    "Number of repeated OrderIDs:",
    df["OrderID"].duplicated().sum()
)

print("\nAll records associated with repeated OrderIDs:")

duplicate_ids = df[
    df["OrderID"].duplicated(keep=False)
]

display(duplicate_ids)
```

```python
df["UnitPrice"] = pd.to_numeric(
    df["UnitPrice"],
    errors="coerce"
)

top_5_products = df.nlargest(5, "UnitPrice")[
    ["Product", "UnitPrice"]
]

print("Five highest-priced products:")
display(top_5_products)
```

```python
print("Highest UnitPrice:", df["UnitPrice"].max())
```

```python
print("1. MISSING VALUES")
print(df.isnull().sum())

print("\n2. DUPLICATE ROWS")
print("Count:", df.duplicated().sum())
display(df[df.duplicated()])

print("\n3. DUPLICATE ORDER IDs")
print("Count:", df["OrderID"].duplicated().sum())
display(df[df["OrderID"].duplicated(keep=False)])

print("\n4. FIVE HIGHEST-PRICED PRODUCTS")
display(df.nlargest(5, "UnitPrice")[["Product", "UnitPrice"]])

print("\n5. HIGHEST UNIT PRICE")
print(df["UnitPrice"].max())
```
