```python
import pandas as pd

df = pd.read_csv("sales_dirty.csv")
```

```python

print("1. FIRST FIVE ROWS")
display(df.head())

print("\n2. DATASET SHAPE")
print(df.shape)

print("\n3. COLUMN NAMES")
print(df.columns.tolist())

print("\n4. DATA TYPES")
print(df.dtypes)

print("\n5. MISSING VALUES")
print(df.isnull().sum())

print("\n6. STATISTICAL SUMMARY")
display(df.describe(include="all"))

print("\n7. EXACT DUPLICATE COUNT")
print(df.duplicated().sum())
```

```python

print("Duplicate records:")
display(df[df.duplicated()])

df = df.drop_duplicates().copy()

print("Rows after removing duplicates:", len(df))
```

```python

df["Product"] = df["Product"].str.strip().str.title()
df["City"] = df["City"].str.strip().str.title()

print("Standardized product names:")
print(df["Product"].unique())

print("\nStandardized city names:")
print(df["City"].unique())
```

```python

df["OrderDate"] = pd.to_datetime(
    df["OrderDate"],
    errors="coerce"
)

print("OrderDate data type:")
print(df["OrderDate"].dtype)

print("\nInvalid or missing dates:")
display(df[df["OrderDate"].isna()])
```

```python

df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

df["UnitPrice"] = pd.to_numeric(
    df["UnitPrice"],
    errors="coerce"
)

print("Numeric column data types:")
print(df[["Quantity", "UnitPrice"]].dtypes)

print("\nMissing or invalid quantities:")
display(df[df["Quantity"].isna()])

print("\nMissing or invalid unit prices:")
display(df[df["UnitPrice"].isna()])
```

```python

quantity_median = df["Quantity"].median()
price_median = df["UnitPrice"].median()

print("Quantity median:", quantity_median)
print("UnitPrice median:", price_median)

df["Quantity"] = df["Quantity"].fillna(quantity_median)
df["UnitPrice"] = df["UnitPrice"].fillna(price_median)
df["City"] = df["City"].fillna("Unknown")

print("\nMissing values after handling:")
print(df.isnull().sum())
```

```python

def categorize_price(price):
    if pd.isna(price):
        return "Unknown"
    elif price < 1000:
        return "Low"
    elif price <= 10000:
        return "Medium"
    else:
        return "High"

df["PriceCategory"] = df["UnitPrice"].apply(categorize_price)

display(df[["Product", "UnitPrice", "PriceCategory"]])
```

```python

df["Revenue"] = df["Quantity"] * df["UnitPrice"]

display(
    df[["OrderID", "Product", "Quantity", "UnitPrice", "Revenue"]]
)

print("Total revenue:", df["Revenue"].sum())
```

```python

q1 = df["UnitPrice"].quantile(0.25)
q3 = df["UnitPrice"].quantile(0.75)

iqr = q3 - q1
upper_limit = q3 + 1.5 * iqr

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Upper outlier limit:", upper_limit)

outliers = df[df["UnitPrice"] > upper_limit]

print("\nPotential price outliers:")
display(outliers[["OrderID", "Product", "UnitPrice"]])
```

```python

print("CLEANED DATASET")
print("Total records:", len(df))
print("Total columns:", len(df.columns))

display(df)
```

```python

print("Missing values:")
print(df.isnull().sum())

print("\nData types:")
print(df.dtypes)

print("\nPrice category counts:")
print(df["PriceCategory"].value_counts())

print("\nTotal revenue:", df["Revenue"].sum())
```
