```python
import numpy as np
import pandas as pd

print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

```python
temperatures = np.array([
    [30, 32, 31, 29, 33],
    [28, 29, 30, 27, 31],
    [35, 34, 36, 33, 37]
])

print("Temperature Array:")
print(temperatures)
```

```python
print("Dimension:", temperatures.ndim)
print("Array Shape:", temperatures.shape)
print("Total Elements:", temperatures.size)
```

```python
print("Total Temperature:", np.sum(temperatures))
print("Average Temperature:", np.mean(temperatures))
print("Maximum Temperature:", np.max(temperatures))
print("Minimum Temperature:", np.min(temperatures))
```

```python
data = {
    'Student': ['Arun', 'Bala', 'Charan', 'Divya', 'Eswar',
                'Farah', 'Gokul', 'Hari'],
    'Maths': [85, 90, 78, 92, 88, 76, 95, 80],
    'Science': [80, 85, 75, 95, 90, 78, 92, 82],
    'English': [88, 92, 80, 89, 85, 75, 94, 84]
}
df = pd.DataFrame(data)
```

```python
df['Total'] = df[['Maths', 'Science', 'English']].sum(axis=1)
```

```python
print("HEAD:")
print(df.head())
```

```python
print("\nTAIL:")
print(df.tail())
```

```python
print("\nSHAPE:")
print(df.shape)
```

```python
print("\nCOLUMNS:")
print(df.columns)
```

```python
print("\nINFO:")
df.info()
```

```python
print("\nNULL VALUES:")
print(df.isnull().sum())
```

```python
print("\nNULL VALUES:")
print(df.isnull().count())
```

```python
print("\nDESCRIPTIVE STATISTICS:")
print(df.describe())
```

```python
print("\nDUPLICATE ROWS:")
print(df.duplicated().sum())
```
