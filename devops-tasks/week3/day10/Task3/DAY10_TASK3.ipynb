```python
import time

scores = [
    [10, 60, 30, 80, 20],
    [70, 40, 90, 50, 15],
    [25, 85, 45, 65, 35],
    [95, 55, 5, 75, 100],
    [20, 30, 60, 40, 80]
]

threshold = 50

start = time.perf_counter()

filtered = [
    x
    for row in scores
    for x in row
    if x > threshold
]

total = sum(filtered)
average = total / len(filtered)

elapsed = time.perf_counter() - start

print("Filtered values:", filtered)
print("Sum:", total)
print("Average:", average)
print("Execution time:", elapsed)
```

```python
import numpy as np
import time

data = np.array([
    [10, 60, 30, 80, 20],
    [70, 40, 90, 50, 15],
    [25, 85, 45, 65, 35],
    [95, 55, 5, 75, 100],
    [20, 30, 60, 40, 80]
])

threshold = 50

start = time.perf_counter()

filtered = data[data > threshold]

total = np.sum(filtered)
average = np.mean(filtered)

elapsed = time.perf_counter() - start

print("Filtered values:", filtered)
print("Sum:", total)
print("Average:", average)
print("Execution time:", elapsed)
```

```python
import timeit

scores = [
    [
        [10, 60, 30],
        [80, 20, 70],
        [40, 90, 50]
    ],
    [
        [15, 85, 45],
        [65, 35, 95],
        [55, 5, 75]
    ],
    [
        [100, 20, 30],
        [60, 40, 80],
        [25, 70, 50]
    ]
]

threshold = 50

def process_array():
    filtered = [
        x
        for layer in scores
        for row in layer
        for x in row
        if x > threshold
    ]

    total = sum(filtered)
    average = total / len(filtered)

    return filtered, total, average

execution_time = timeit.timeit(
    process_array,
    number=10000
) / 10000

filtered, total, average = process_array()

print("Filtered values:", filtered)
print("Sum:", total)
print("Average:", average)
print("Average execution time:", execution_time, "seconds")
```
